```python
import pandas as pd
import numpy as np
import re

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder

import joblib 
```

```python
career_df = pd.read_csv('career_it_recomendation.csv')

career_df.head()
```

# 3. REMOVE INVALID TARGET

```python
remove_jobs = [
    "Tidak Tersedia",
    "Mahasiswa (Belum Bekerja)"
]

career_df = career_df[
    ~career_df['pekerjaan_pertama'].isin(remove_jobs)
].copy()
```

# 4. DROP DUPLICATE

```python
career_df = career_df.drop_duplicates().copy()
```

# 5. feature & target

```python
X = career_df['profil_lengkap']
y = career_df['pekerjaan_pertama']
```

# 6. Text Cleaning

```python
def clean_text(text):
    
    text = str(text).lower()
    
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
    
    text = re.sub(r'\s+', ' ', text)
    
    return text.strip()
```

```python
X_cleaned = X.apply(clean_text)
```

### Text Cleaning

Tahap preprocessing dilakukan untuk membersihkan teks dari karakter khusus, simbol, dan inkonsistensi penulisan agar data lebih siap diproses oleh model machine learning.

# 7. TF-IDF Vectorization

```python
tfidf = TfidfVectorizer(max_features=1000)
```

```python
X_tfidf = tfidf.fit_transform(X_cleaned)
```

```python
X_tfidf.shape
```

### TF-IDF Feature Engineering

Data teks diubah menjadi representasi numerik menggunakan TF-IDF agar model machine learning dapat memahami pola kata dan keterampilan yang muncul pada profil pengguna.

# 8. Label Encoding Target

```python
encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)
```

```python
encoder.classes_
```

# 9. Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y_encoded, test_size=0.2, random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

# 10. Save Preprocessing Artifact

```python
import os

os.makedirs("../models", exist_ok=True)
```

```python
joblib.dump(tfidf, "../models/tfidf_vectorizer_it.pkl")
joblib.dump(encoder, "../models/label_encoder_it.pkl")
```
